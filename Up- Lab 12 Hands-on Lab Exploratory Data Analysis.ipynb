#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.


```python
# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Set pandas option to display all columns
pd.set_option('display.max_columns', None)

# Display the first few rows of the dataset
df.head()
```

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
def safe_mode(series):
    modes = series.mode()
    return modes.iloc[0] if len(modes)>0 else "Unknown"

df["Employment"] = df["Employment"].fillna(safe_mode(df["Employment"]))
df["JobSat"] = df["JobSat"].fillna(safe_mode(df["JobSat"]))
df["RemoteWork"] = df["RemoteWork"].fillna(safe_mode(df["RemoteWork"]))

missing_values = df[["Employment", "JobSat", "RemoteWork"]].isnull().sum()
print("步驟3 處理後缺失：")
print(missing_values)
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
def clean_yearspro(val):
    if pd.isna(val):
        return None
    val = str(val).strip()
    if val == "Less than 1 year":
        return 0
    if val == "More than 50 years":
        return 51
    try:
        return int(val)
    except:
        return None

df["YearsCodePro_clean"] = df["YearsCodePro"].apply(clean_yearspro)

df["JobSat_Num"] = df["JobSat"]

bins = [0, 5, 10, 20, float("inf")]
labels = ["0‑5", "5‑10", "10‑20", ">20"]
df["ExperienceRange"] = pd.cut(df["YearsCodePro_clean"], bins=bins, labels=labels, right=False)

median_Jobsat = df.groupby("ExperienceRange")["JobSat_Num"].median()
print("各經驗區間滿意度中位數：")
print(median_Jobsat)

plt.figure(figsize=(8,6))
plt.bar(median_Jobsat.index.astype(str), median_Jobsat.values, color="steelblue", edgecolor="black")
plt.xlabel("Years of Professional Coding")
plt.ylabel("Median Job Satisfaction (0‑10 scale)")
plt.title("Professional Coding Experience vs Median Job Satisfaction")
plt.ylim(0,11)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
plt.figure(figsize=(8, 6))
sns.countplot(data=df, x="JobSat", color="#4472C4")

plt.xlabel("JobSat Level")
plt.ylabel("Count of Respondents")
plt.title("Distribution of Job Satisfaction")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
plt.figure(figsize=(8, 6))
sns.countplot(data=df, x="RemoteWork", order=df["RemoteWork"].value_counts().index)
plt.xlabel("Remote Work Status")
plt.ylabel("Count of Respondents")
plt.title("Distribution of Remote Work")
plt.xticks(rotation=30, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

cross_tab = pd.crosstab(df["Employment"], df["RemoteWork"], normalize="index") * 100
print("Remote Work Preferences by Employment Type (%):")
print(cross_tab.round(1))

top_dev = df["DevType"].value_counts().head(6).index
df_dev_filter = df[df["DevType"].isin(top_dev)]

plt.figure(figsize=(10, 6))
sns.countplot(data=df_dev_filter, y="DevType", hue="RemoteWork")
plt.xlabel("Count of Respondents")
plt.ylabel("Developer Role")
plt.title("Remote Work Preference by Developer Role")
plt.tight_layout()
plt.show()

dev_cross_tab = pd.crosstab(df_dev_filter["DevType"], df_dev_filter["RemoteWork"], normalize="index") * 100
print("\nRemote Work Preferences by Developer Role (%):")
print(dev_cross_tab.round(1))
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
lang_df = df.copy()
lang_df = lang_df.assign(LanguageHaveWorkedWith = lang_df["LanguageHaveWorkedWith"].str.split(";")).explode("LanguageHaveWorkedWith")
lang_df = lang_df.dropna(subset=["Country", "LanguageHaveWorkedWith"])

lang_df = lang_df.reset_index(drop=True)

top_countries = lang_df["Country"].value_counts().head(5).index
lang_top_country = lang_df[lang_df["Country"].isin(top_countries)].reset_index(drop=True)

country_lang_count = lang_top_country.groupby(["Country","LanguageHaveWorkedWith"]).size().reset_index(name="count")

top_language_per_country = country_lang_count.sort_values(["Country","count"],ascending=False).groupby("Country").head(6)

plt.figure(figsize=(12,7))
sns.barplot(data=top_language_per_country, x="count", y="LanguageHaveWorkedWith", hue="Country")
plt.xlabel("Number of Respondents")
plt.ylabel("Programming Language")
plt.title("Most Popular Programming Languages by Country")
plt.tight_layout()
plt.show()

pivot_table = country_lang_count.pivot_table(index="Country", columns="LanguageHaveWorkedWith", values="count", fill_value=0)

plt.figure(figsize=(14,6))
sns.heatmap(pivot_table, cmap="Blues")
plt.xlabel("Programming Language")
plt.ylabel("Country")
plt.title("Programming Language Usage Heatmap by Region")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

print("=== Country vs Programming Language Count ===")
print(country_lang_count.head(12))
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
def clean_yearspro(val):
    if pd.isna(val):
        return None
    val = str(val).strip()
    if val == "Less than 1 year":
        return 0
    if val == "More than 50 years":
        return 51
    try:
        return int(val)
    except:
        return None

df["YearsCodePro_clean"] = df["YearsCodePro"].apply(clean_yearspro)

df_scatter = df.dropna(subset=["YearsCodePro_clean", "JobSatPoints_1"])

plt.figure(figsize=(10,6))
sns.scatterplot(data=df_scatter, x="YearsCodePro_clean", y="JobSatPoints_1", alpha=0.2, s=15)

plt.xlabel("Professional Coding Experience (Years)")
plt.ylabel("JobSatPoints_1 (Satisfaction Score)")
plt.title("Relationship between Professional Experience and Job Satisfaction")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

corr_value = df_scatter["YearsCodePro_clean"].corr(df_scatter["JobSatPoints_1"])
print(f"Correlation (YearsCodePro vs JobSatPoints_1): {corr_value:.3f}")
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
df_emp = df.copy()
df_emp = df_emp.assign(Employment = df_emp["Employment"].str.split(";")).explode("Employment")
df_emp = df_emp.dropna(subset=["EdLevel","Employment"]).reset_index(drop=True)

cross_ed_emp = pd.crosstab(df_emp["EdLevel"], df_emp["Employment"])
print("=== EdLevel vs Employment 原始計數 ===")
print(cross_ed_emp)

cross_ed_emp_pct = pd.crosstab(df_emp["EdLevel"], df_emp["Employment"], normalize="index") * 100
print("\n=== EdLevel vs Employment 行百分比(%) ===")
print(cross_ed_emp_pct.round(1))

plt.figure(figsize=(14,7))
cross_ed_emp_pct.plot(kind="bar", stacked=True, ax=plt.gca(), cmap="Set2")

plt.xlabel("Education Level (EdLevel)")
plt.ylabel("Percentage (%)")
plt.title("Employment Type Distribution by Education Level")
plt.xticks(rotation=40, ha="right")

plt.legend(title="Employment Type", bbox_to_anchor=(1.02, 1), loc="upper left", fontsize=8)
plt.subplots_adjust(right=0.80)
plt.show()

plt.figure(figsize=(14,6))
sns.heatmap(cross_ed_emp, annot=True, fmt="d", cmap="YlGnBu")
plt.xlabel("Employment Type")
plt.ylabel("Education Level")
plt.title("Heatmap: Education Level vs Employment Type")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
