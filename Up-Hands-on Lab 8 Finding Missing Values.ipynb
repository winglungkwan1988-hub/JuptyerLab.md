##### Import Necessary Modules:


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Tasks


<h2>1. Load the Dataset</h2>
<p>
We use the <code>pandas.read_csv()</code> function for reading CSV files. However, in this version of the lab, which operates on JupyterLite, the dataset needs to be downloaded to the interface using the provided code below.
</p>


The functions below will download the dataset into your browser:



```python
# Define the URL of the dataset
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### 2. Explore the Dataset
##### Task 1: Display basic information and summary statistics of the dataset.


```python
df.info()
df.describe()
print(df.isnull().sum())
print(df.isnull().mean() * 100)
print(df[df.isnull().any(axis=1)])
```

### 3. Finding Missing Values
##### Task 2: Identify missing values for all columns.


```python
missing_values = df.isnull().sum()
print("Missing values per column:")
print(missing_values)
```

##### Task 3: Visualize missing values using a heatmap (Using seaborn library).



```python
plt.figure(figsize=(12, 6))

sns.heatmap(df.isnull(), cbar=False, cmap="viridis", yticklabels=False)
plt.title("Missing Values Heatmap")
plt.xlabel("Columns")
plt.ylabel("Rows")
plt.show()
```

##### Task 4: Count the number of missing rows for a specific column (e.g., `Employment`).


```python
missing_employment = df["Employment"].isnull().sum()
print("Number of missing rows in Employment column:", missing_employment)
```

### 4. Imputing Missing Values
##### Task 5: Identify the most frequent (majority) value in a specific column (e.g., `Employment`).


```python
majority_value = df["Employment"].mode()[0]
print(majority_value)
```

##### Task 6: Impute missing values in the `Employment` column with the most frequent value.



```python
majority_value = df["Employment"].mode()[0]
print("多数值:", majority_value)
df["Employment"] = df["Employment"].fillna(majority_value)
print("填补后缺失数:", df["Employment"].isnull().sum())
```

### 5. Visualizing Imputed Data
##### Task 7: Visualize the distribution of a column after imputation (e.g., `Employment`).


```python
top9 = df["Employment"].value_counts().head(9).index
df["Employment_grouped"] = df["Employment"].where(
    df["Employment"].isin(top9), 
    "Other"
)

plt.figure(figsize=(12, 6))
sns.countplot(
    data=df, 
    x="Employment_grouped",
    order=df["Employment_grouped"].value_counts().index
)
plt.xticks(rotation=45, ha="right")
plt.title("Employment Distribution (Grouped)")
plt.tight_layout()
plt.show()
```
