### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**
<p>
The functions below will download the dataset into your browser:
</p>


```python
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
duplicates = df.duplicated()
print("Number of duplicate rows:", duplicates.sum())
print("Sample duplicate rows:")
print(df[duplicates].head())
```

##### Task 2: Remove the duplicate rows from the dataframe.



```python
df_cleaned = df.drop_duplicates()
print("Original shape:", df.shape)
print("Cleaned shape:", df_cleaned.shape)
print("Rows removed:", df.shape[0] - df_cleaned.shape[0])
print("Duplicates remaining:", df_cleaned.duplicated().sum())
```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
missing_value = df_cleaned.isnull().sum()
print(missing_value)
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
missing_remotework = df_cleaned["RemoteWork"].isnull().sum()
print(missing_remotework)
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
Counts = df_cleaned["RemoteWork"].value_counts()
print(Counts)
```

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
majority_value = df_cleaned["RemoteWork"].mode()[0]
print(majority_value)
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
df_cleaned["RemoteWork"] = df_cleaned["RemoteWork"].fillna(majority_value)
print("Missing values after filling:", df_cleaned["RemoteWork"].isnull().sum())
print(df_cleaned["RemoteWork"])
```

##### Task 8: Check for any compensation-related columns and describe their distribution.



```python
comp_columns = [col for col in df_cleaned.columns if "comp" in col.lower()]
print("Compensation-related columns:", comp_columns)
print(df_cleaned[comp_columns].describe(include='all'))
```
