### Task 1: Visualizing Data Composition with Pie Charts


##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



```python
df_counts = df["DatabaseWantToWorkWith"].value_counts()
df_clean = df_counts.dropna()
top5 = df_clean.head(5)

plt.figure(figsize=(8,6))
plt.pie(top5.values, labels=top5.index, autopct="%1.1f%%")
plt.title("Pie Chart of Top 5 Database")
plt.show()
```

The `DevType` column lists the developer types for respondents. We’ll examine the distribution by showing the top 5 developer roles in a pie chart.



```python
df_clean = df["DevType"].dropna()
df_dev = df_clean.value_counts()
top5 = df_dev.head(5)
plt.figure(figsize=(8,6))
plt.pie(top5.values, labels=top5.index, autopct="%1.1f%%")
plt.title("Pie Chart of Top 5 Developer Roles")
plt.show()
```

##### 1.3 Create a pie chart for the operating systems used by respondents for professional use


The `OpSysProfessional` use column shows the operating systems developers use professionally. Let’s visualize the distribution of the top operating systems in a pie chart.



```python
mode_value = df["OpSysProfessional use"].mode()[0]
df["OpSysProfessional use"].fillna(mode_value)

os_counts = df["OpSysProfessional use"].value_counts()
top5 = os_counts.head(5)
others = os_counts.iloc[5:].sum()
top5_with_others = pd.concat([top5, pd.Series({"Others": others})])
plt.pie(top5_with_others.values, labels=top5_with_others.index, autopct="%1.1f%%")
plt.show()
```

### Task 2: Additional Visualizations and Comparisons


##### 2.1 Pie Chart for Top 5 Programming Languages Respondents Have Worked With


The `LanguageHaveWorkedWith` column contains the programming languages that respondents have experience with. We’ll plot a pie chart to display the composition of the top 5 languages.



```python
print(df["LanguageHaveWorkedWith"])
missing = df["LanguageHaveWorkedWith"].isnull().mean() * 100
print(missing)

df_clean = df["LanguageHaveWorkedWith"].dropna()
df_lang = df_clean.value_counts()
top5 = df_lang.head(5)
plt.figure(figsize=(8,6))
plt.pie(top5.values, labels=top5.index, autopct="%1.1f%%")
plt.title("Top 5 Language")
plt.show()
```

##### 2.2 Pie Chart for Top Collaboration Tools used in Professional Use


Using the `NEWCollabToolsHaveWorkedWith` column, we’ll identify and visualize the top collaboration tools respondents use in their professional work.



```python
print(df["NEWCollabToolsHaveWorkedWith"])
missing = df["NEWCollabToolsHaveWorkedWith"].isnull().mean() * 100
print(missing)
mode_value = df["NEWCollabToolsHaveWorkedWith"].mode()[0]
df["NEWCollabToolsHaveWorkedWith"].fillna(mode_value)
df_tool = df["NEWCollabToolsHaveWorkedWith"].value_counts()
top5 = df_tool.head(5)
others = df_tool.iloc[5:].sum()
top5_with_others = pd.concat([top5, pd.Series({"Others": others})])
plt.figure(figsize=(8,6))
plt.pie(top5_with_others.values, labels=top5_with_others.index, autopct="%1.1f%%")
plt.title("Top 5 Collaboration Tools")
plt.show()
```

### Task 3: Analyzing and Interpreting Composition


In this task, you will create additional pie charts to analyze specific aspects of the survey data. Use `pandas` and `matplotlib` to complete each task and interpret the findings.



##### 3.1 Pie Chart of `Respondents` Most Admired Programming Languages


The `LanguageAdmired` column lists the programming languages respondents admire most. Create a pie chart to visualize the top 5 admired languages.



```python
print(df["LanguageAdmired"])
missing = df["LanguageAdmired"].isnull().mean() * 100
print(missing)
mode_value = df["LanguageAdmired"].mode()[0]
df["LanguageAdmired"].fillna(mode_value)
df_lang = df["LanguageAdmired"].value_counts()
top5 = df_lang.head(5)
plt.figure(figsize=(8,6))
plt.pie(top5.values, labels=top5.index, autopct="%1.1f%%")
plt.title("Pie Chart of Respondents Most Admired Programming Languages Top 5")
plt.show()
```

##### 3.2 Pie Chart of Tools Used for AI Development


Using the `AIToolCurrently` Using column, create a pie chart to visualize the top 5 tools developers are currently using for AI development.



```python
print(df["AIToolCurrently Using"])
missing = df["AIToolCurrently Using"].isnull().mean() * 100
print(missing)

df["AIToolCurrently Using"] = df["AIToolCurrently Using"].fillna("Unknown")
df_exploded = df["AIToolCurrently Using"].str.split(";").explode()
df_aitool = df_exploded.value_counts()
top5 = df_aitool.head(6)
plt.figure(figsize=(8,6))
plt.pie(top5.values, labels=top5.index, autopct="%1.1f%%")
plt.title("Pie Chart of Top 5 Tools AI Development")
plt.show()
```

##### 3.3 Pie Chart for Preferred Web Frameworks


The `WebframeWantToWorkWith` column includes web frameworks that respondents are interested in working with. Visualize the top 5 frameworks in a pie chart.



```python
print(df["WebframeWantToWorkWith"])
missing = df["WebframeWantToWorkWith"].isnull().mean() * 100
print(missing)

df["WebframeWantToWorkWith"] = df["WebframeWantToWorkWith"].fillna("Unknown")
df_exploded = df["WebframeWantToWorkWith"].str.split(";").explode()
df_web = df_exploded.value_counts()
top5 = df_web.head(6)
plt.figure(figsize=(8,6))
plt.pie(top5.values, labels=top5.index, autopct="%1.1f%%")
plt.title("Pie Chart of Top 5 Preferred Web Frameworks")
plt.show()
```

##### 3.4 Pie Chart for Most Desired Embedded Technologies


Using the `EmbeddedWantToWorkWith` column, create a pie chart to show the top 5 most desired embedded technologies that respondents wish to work with.



```python
print(df["EmbeddedWantToWorkWith"])
missing = df["EmbeddedWantToWorkWith"].isnull().mean() * 100
print(missing)

df["EmbeddedWantToWorkWith"] = df["EmbeddedWantToWorkWith"].fillna("Unknown")
df_exploded = df["EmbeddedWantToWorkWith"].str.split(";").explode()
df_tech = df_exploded.value_counts()
top5 = df_tech.head(6)
plt.figure(figsize=(8,6))
plt.pie(top5.values, labels=top5.index, autopct="%1.1f%%")
plt.title("Pie Chart of Top 5 Most Desired Embedded Technologies")
plt.show()
```
