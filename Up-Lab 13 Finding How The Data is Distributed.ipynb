### Step 1: Import Libraries and Load Data


- Import the `pandas`, `matplotlib.pyplot`, and `seaborn` libraries.


- You will begin with loading the dataset. You can use the pyfetch method if working on JupyterLite. Otherwise, you can use pandas' read_csv() function directly on their local machines or cloud environments.


```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Display the first few rows of the dataset
df.head()
```

### Step 2: Examine the Structure of the Data


- Display the column names, data types, and summary information to understand the data structure.

- Objective: Gain insights into the dataset's shape and available variables.


```python
print(df.columns)
print(df.dtypes)
df.info()
df.describe(include="all")
```

### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
remaining_cols = df.columns[df.isnull().any()].tolist()
print("剩余缺失列的数据类型分布:")
print(df[remaining_cols].dtypes.value_counts())

if 'YearsCodePro' in df.columns and df['YearsCodePro'].isnull().sum() > 0:
    df['YearsCodePro'] = df['YearsCodePro'].replace({
        'Less than 1 year': 0.5,
        'More than 50 years': 50
    })
    df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')
    df['YearsCodePro'] = df['YearsCodePro'].fillna(df['YearsCodePro'].median())
    print("已处理 YearsCodePro")

numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
numeric_cols = [c for c in numeric_cols if c != 'ConvertedCompYearly' and c != 'ResponseId']

for col in numeric_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

print(f"已用中位数填充 {len(numeric_cols)} 个数值列")

categorical_cols = df.select_dtypes(include=['object', 'str']).columns

for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna('Not Specified')

print(f"已用 'Not Specified' 填充 {len(categorical_cols)} 个分类列")

remaining_missing = df.isnull().sum().sum()
print(f"\n=== 最终验证 ===")
print(f"剩余缺失值总数: {remaining_missing}")

if remaining_missing > 0:
    remaining = df.isnull().sum()
    remaining = remaining[remaining > 0]
    print("仍有缺失的列:")
    print(remaining)
else:
    print("除 ConvertedCompYearly 外，所有缺失值已处理完毕！")

print(f"\nConvertedCompYearly 缺失数（故意保留）: {df['ConvertedCompYearly'].isnull().sum()}")
print(f"SalaryProvided 标记列已创建: {'SalaryProvided' in df.columns}")
print(f"\n最终数据形状: {df.shape}")
```

### Step 4: Analyze Key Columns


- Examine key columns such as `Employment`, `JobSat` (Job Satisfaction), and `YearsCodePro` (Professional Coding Experience).

- **Instruction**: Calculate the value counts for each column to understand the distribution of responses.



```python
print("Employment distribution:")
print(df["Employment"].value_counts())
print(df["Employment"].value_counts(normalize=True).round(3) * 100)

print("\nJob Satisfaction distribution:")
print(df["JobSat"].describe())
print(df["JobSat"].value_counts().sort_index())

print("\nYears of Professional Coding Experience distribution:")
print(df["YearsCodePro"].describe())
print(df["YearsCodePro"].value_counts().sort_index())
```

### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
plt.figure(figsize=(10, 5))
sns.kdeplot(data=df, x='JobSat', fill=True, color='steelblue', linewidth=2)
plt.axvline(df['JobSat'].mean(), color='red', linestyle='--', label=f'Mean: {df["JobSat"].mean():.2f}')
plt.axvline(df['JobSat'].median(), color='green', linestyle='--', label=f'Median: {df["JobSat"].median():.1f}')
plt.title('Job Satisfaction Distribution (KDE)')
plt.xlabel('Job Satisfaction Score (0-10)')
plt.ylabel('Density')
plt.legend()
plt.xlim(0, 10)
plt.tight_layout()
plt.show()
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
def count_languages_pd(series, col_name):
    """用 str.get_dummies 拆分分号分隔的语言列表，统计频率"""
    filtered = series[series != 'Not Specified'].fillna('')
    dummies = filtered.str.get_dummies(sep=';')
    counts = dummies.sum().sort_values(ascending=False)
    result = counts.reset_index()
    result.columns = ['Language', col_name]
    return result

have_df = count_languages_pd(df['LanguageHaveWorkedWith'], 'HaveWorked')
want_df = count_languages_pd(df['LanguageWantToWorkWith'], 'WantToWork')

lang_compare = pd.merge(have_df, want_df, on='Language', how='outer').fillna(0)
lang_compare['HaveWorked'] = lang_compare['HaveWorked'].astype(int)
lang_compare['WantToWork'] = lang_compare['WantToWork'].astype(int)

lang_compare['NetChange'] = lang_compare['WantToWork'] - lang_compare['HaveWorked']
lang_compare['Want_Ratio'] = (lang_compare['WantToWork'] / lang_compare['HaveWorked']).round(2)

print("=== 编程语言对比（Top 20）===")
print(lang_compare.sort_values('HaveWorked', ascending=False).head(20).to_string(index=False))

top15 = lang_compare.sort_values('HaveWorked', ascending=False).head(15)

plt.figure(figsize=(14, 7))
x = range(len(top15))
width = 0.35

plt.bar([i - width/2 for i in x], top15['HaveWorked'], width,
        label='Have Worked With', color='steelblue', alpha=0.8)
plt.bar([i + width/2 for i in x], top15['WantToWork'], width,
        label='Want To Work With', color='coral', alpha=0.8)

plt.xticks(x, top15['Language'], rotation=45, ha='right')
plt.ylabel('Number of Respondents')
plt.title('Top 15 Programming Languages: Have Worked vs Want To Work')
plt.legend()
plt.tight_layout()
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
print("=== RemoteWork 唯一值 ===")
print(df['RemoteWork'].value_counts())

print("\n=== Country 列是否存在 ===")
if 'Country' in df.columns:
    print(f"存在，共 {df['Country'].nunique()} 个国家/地区")
    print(df['Country'].value_counts().head(10))
else:
    print("不存在 Country 列，请检查可用的地区列：")
    print([c for c in df.columns if 'count' in c.lower() or 'region' in c.lower() or 'loc' in c.lower()])

df_remote = df[(df['RemoteWork'] != 'Not Specified') & 
               (df['Country'] != 'Not Specified')].copy()

top_countries = df_remote['Country'].value_counts().head(15).index.tolist()
df_top = df_remote[df_remote['Country'].isin(top_countries)]

cross = pd.crosstab(df_top['Country'], df_top['RemoteWork'])

cross_pct = cross.div(cross.sum(axis=1), axis=0) * 100

print("=== 各国家远程工作分布（人数）===")
print(cross)
print("\n=== 各国家远程工作分布（百分比）===")
print(cross_pct.round(1))

cross_pct.plot(kind='bar', figsize=(14, 7), 
               color=['#66b3ff', '#99ff99', '#ff9999'],
               edgecolor='white')
plt.title('Remote Work Distribution by Country (Top 15)')
plt.xlabel('Country')
plt.ylabel('Percentage (%)')
plt.legend(title='Remote Work Type', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

global_remote = df_remote['RemoteWork'].value_counts()
global_pct = (global_remote / global_remote.sum() * 100).round(1)

print("\n=== 全球整体远程工作分布 ===")
print(pd.DataFrame({'人数': global_remote, '占比(%)': global_pct}))

plt.figure(figsize=(8, 5))
bars = sns.barplot(x=global_pct.index, y=global_pct.values, 
                    palette=['#66b3ff', '#99ff99', '#ff9999'])
plt.title('Global Remote Work Distribution')
plt.xlabel('Remote Work Type')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=20, ha='right')

for i, v in enumerate(global_pct.values):
    bars.text(i, v + 0.5, f'{v}%', ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
pearson_corr = df["YearsCodePro"].corr(df["JobSat"], method="pearson")
print("Pearson correlation:", pearson_corr)
```

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
top_emp = df['Employment'].value_counts().head(5).index.tolist()
df['EmpGroup'] = df['Employment'].where(df['Employment'].isin(top_emp), 'Other')

cross_tab = pd.crosstab(df['EdLevel'], df['EmpGroup'], normalize='index') * 100
print(cross_tab.round(1))

cross_tab.plot(kind='bar', stacked=True, figsize=(12, 6), colormap='Set3')
plt.title('Employment Type Distribution by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Percentage (%)')
plt.legend(title='Employment Type', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
